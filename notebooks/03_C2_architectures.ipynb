# 03 — Contribution 2: Architectural ablation

This notebook evaluates alternative GNN backbones and fusion modules, keeping the semantic embedding fixed.

## Configurations evaluated (FastText embedding)
- GraphSAGE + Multi-Head Attention   ← **C2 winner**
- GCN + Multi-Head Attention
- GAT + Pooling
- GAT + Averaging

The GAT + MHA (FastText) baseline is reported in `02_C1_embeddings.ipynb`.

## Cross-check (Word2Vec embedding)
- Word2Vec + GraphSAGE + MHA   ← **Top absolute F1**

## Expected results
| GNN | Fusion | F1 |
|---|---|---|
| GAT | MHA (baseline) | ≈ 0.9725 |
| **GraphSAGE** | **MHA** | **≈ 0.9932** |
| GCN | MHA | ≈ 0.9540 |
| GAT | Pooling | ≈ 0.9254 |
| GAT | Averaging | ≈ 0.9292 |
| W2V + **GraphSAGE** | **MHA** | **≈ 0.9933** |

- `02_C1_embeddings.ipynb` completed.
- Trained weights present under `trained_weights/trace/{c2_*, c2_w2v_*}/`.
